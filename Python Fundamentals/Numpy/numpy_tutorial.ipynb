```python
import numpy as np
import time as time 
import sys
```

```python
size = 1_000_000

# Python list
py_list = list(range(size))
start = time.time()
sq_list = [x**2 for x in py_list]
end = time.time()
print(f"The python list was executed in {end - start} time")

## Numpy array
np_array = np.array(range(size))
start = time.time()
sq_array = np_array ** 2 
end = time.time()
print(f"The numpy array was executed in {end - start} time")
```

```python
## Memomry 
print(f"The size of the python list is {sys.getsizeof(py_list) * len(py_list)}")
print(f"The size of the numpy array is {np_array.nbytes}")
```
