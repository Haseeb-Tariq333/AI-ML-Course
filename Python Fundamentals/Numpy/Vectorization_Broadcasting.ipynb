```python
import numpy as np
```

```python
## Vectorization ## 

arr = np.array([1,2,3,4,5])
arr2 = np.array([6,7,8,9,10])

print(arr + arr2) ## ADD TWO ARRAYS 
print(arr**2) ## TAKE POWER OF 2 FOR EACH ELEMENT IN THE ARRAY
print(arr+2) ## ADD TW0 TO EACH ELEMENT IN THE ARRAY 
```

```python
## Vector Normalization ##
arr = np.array([[1,2,3],[4,5,6]])
mean = np.mean(arr)
std_dev = np.std(arr)
normalized_arr = (arr - mean)/std_dev
print(normalized_arr)
print(mean)
print(std_dev)
```

```python
## Mathematics functions ##

arr = np.array([1,2,3,5,4])

print(np.sum(arr))
print(np.prod(arr))
print(np.min(arr))
print(np.max(arr))
print(np.argmin(arr))
print(np.argmax(arr))
print(np.median(arr))
print(np.std(arr))
print(np.mean(arr))
print(np.var(arr))
print(np.square(arr))
print(np.sqrt(arr))
print(np.pow(arr,3))
print(np.log(arr))
print(np.log10(arr))
print(np.log2(arr))
print(np.sort(arr))
print(np.unique(arr))
print(np.abs(arr))
print(np.round(3,14))
print(np.ceil(3.14)) ## round off to the upper value
print(np.floor(3.14)) ## round of to the lower value
```
