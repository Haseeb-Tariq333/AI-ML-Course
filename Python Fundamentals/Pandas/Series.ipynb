```python
import pandas as pd
```

```python
s = pd.Series([1,2,3,4,5])
print(s)
print(type(s))

## Indexing 
print(s[0])
print(s[3])

```

```python
## Labels ## 
s = pd.Series([23,34,45,56], index = ["Haseeb", "Habib", "Abdullah","Ahmed"])
print(s)
print(s[0])
print(s["Haseeb"])
print(s.index)
```

```python
#Vectoriztion ##
s1 = pd.Series([1,2,3,4,5])
s2 = pd.Series([10,11,12,13,14])
print(s1+s2)

s1[0] = 100
print(s1)

changes_s1 = s1.drop(0)
print(changes_s1)
```
