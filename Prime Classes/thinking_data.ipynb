```python
import json
```

```python
# Load data
def print_data(filename):
    with open(filename , "r") as f:
        data = json.load(f)

    return data
```

```python
data = print_data("store_data.json")
```

```python
def clean_data(data):
    mapping_rating = {"one":1, "two":2, "three":3, "four":4, "five":5}
    cleaned_data = []
    unique_users = set()

    for users in data:
        raw_rating = str(users["rating"]).strip().lower()
        if raw_rating in mapping_rating:
            raw_rating = mapping_rating[raw_rating]
        users["rating"] = raw_rating 

        ## Handling missing data
        raw_age = users.get("age")
        if raw_age == None:
            users[raw_age] = None

        ## Handling duplicated data
        if users["name"].strip() in unique_users:
            continue
        unique_users.add(users["name"])
        cleaned_data.append(users)
    return cleaned_data
        
```

```python
data = clean_data(data)
```

```python
def avg_rating(data):
    tot_rating = 0
    for user in data:
        tot_rating += float(user["rating"])
    print(f"The average rating is {tot_rating/len(data)}")
        
```

```python
avg_rating(data)
```

```python
def poor_rating(data):
    poor_rating = 0
    for users in data:
        if(float(users["rating"]) < 3):
            poor_rating += 1
            
    print(f"The percentage of poor rating is {poor_rating/len(data) * 100}")
```

```python
poor_rating(data)
```

```python
def user_recommendation(data):
    recommendations = []
    for user in data:
        curr_recomm = {}
        curr_recomm["name"] = user["name"]
    
        if float(user["rating"]) >= 4:
            curr_recomm["brand"] = "Apple"
        else:
            curr_recomm["brand"] = "Samsung"

        recommendations.append(curr_recomm)
    return recommendations
```

```python
print(user_recommendation(data))
```
