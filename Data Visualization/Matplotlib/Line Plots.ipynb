```python
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The Kings Speech",
    "The Artist",
    "Argo"
]
years = [2008, 2009,2010,2011,2012]
oscar_revenue = [1005,170,427,133,232] # in M$

plt.plot(years, oscar_revenue)

plt.title("Oscar Movies Revenue in each year")
plt.xlabel("Years")
plt.ylabel("Revenue in M$")
```

```python
non_ocsar_movies = [
    "Slumdog Millionaire",
    "Avatar",
    "Inception",
    "Hugo",
    "Lincoln"
]
non_oscar_revenue = [378,2788,829,185,275]
plt.plot(years, oscar_revenue, "o--r", label="Oscar Movies")
plt.plot(years, non_oscar_revenue, "o:y", label="Non-Oscar Movies")

plt.title("Oscar Movies vs Non Oscar Movies Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue in M$")
plt.legend()

plt.savefig("Line_Plot.png")
```
