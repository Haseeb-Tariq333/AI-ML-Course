```python
import matplotlib.pyplot as plt
```

```python
import numpy as np

np.random.seed(0)
scores = np.random.normal(70,10,100)

bins = [0,30,70,90,100]

plt.hist(scores, edgecolor = "black", bins=bins)
plt.xlabel("Scores")
plt.ylabel("# of students")

```

```python
marks = [
    25,38,42,55,61,67,72,78,83,88,
    91,95,57,53,59,64,69,74,79,85,
    92,97,33,45,58,62,71,76,81,89
]

plt.hist(marks,color = "skyblue", edgecolor = "black", bins= bins)

plt.axvline(
    x=33,
    label = "Passing marks",
    linewidth="2",
    color = "red",
    linestyle="--"
)
plt.xlabel("Marks")
plt.ylabel("Number of students")
```

```python
legit_transactions = [
    2.99, 5.49, 8.99, 12.50, 14.99, 19.99, 23.45, 29.99, 34.99, 39.50,
    45.00, 49.99, 55.25, 60.00, 75.99, 89.99, 120.50, 150.00, 199.99,
    249.99, 300.75, 450.00, 600.00, 850.00, 1200.00
]

fraud_transactions = [
    50, 100, 150, 200, 300, 500, 500, 750, 1000, 1000,
    1200, 1500, 1500, 2000, 2500, 3000, 3000
]

plt.hist(legit_transactions, color="green", bins = 10, edgecolor="black", label="Legit", alpha = 0.5 )
plt.hist(fraud_transactions, color="red", bins=10, edgecolor="black", label = "fraud", alpha = 0.5)

plt.title("Fraid vs legit transactions")
plt.xlabel("Amount")
plt.ylabel("Frequency of transaction")
plt.legend()
```
