```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.set_theme()
```

```python
flights = sns.load_dataset("flights")
flights
flights_pivot = flights.pivot(index="month", columns="year", values="passengers")
```

```python
sns.heatmap(
    data = flights_pivot,
    cmap = "coolwarm",
    annot = True,
    fmt = "d"
)
```
