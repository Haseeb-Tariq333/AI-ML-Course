```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
sns.set_theme()
```

```python
## Question ! ##
random_data = np.random.normal(loc=70, scale=10, size=200)
mean_val = np.mean(random_data)
# print(random_data)

plt.hist(random_data, bins=20, color = "skyblue", edgecolor = "black", alpha = 0.8)
plt.axvline(
    mean_val,
    color = "red",
    linestyle = "--"
)
```

```python
## QUESTION 2 ##

penguins = sns.load_dataset("penguins")
penguins
```

```python
sns.relplot(
    data = penguins,
    x = "flipper_length_mm",
    y = "body_mass_g",
    hue = "species"
)
```

```python
## QUESTION 3 ##
tips =  sns.load_dataset("tips")
sns.boxplot(
    data = tips,
    x = "total_bill",
    y = "day",
    hue = "sex"
    
)
```

```python
## QUESTION 4 ##
sns.relplot(
    data = tips,
    x = "total_bill",
    y = "tip",
    hue = "sex",
    style = "smoker"
)
```

```python
pivot_table = tips.pivot_table(index="day", columns= "time", values= "tip", aggfunc= "mean", observed= False)
pivot_table
```

```python
sns.heatmap(
    data = pivot_table,
    cmap = "coolwarm",
    annot = True
)
```
