```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.set_theme()
```

```python
sns.get_dataset_names()
```

```python
tips = sns.load_dataset("tips")
print(type(tips))
print(tips.head())
```

```python
## scatter plots

sns.relplot(
    data = tips,
    x = "total_bill",
    y = "tip",
    hue = "day"
)
```

```python
## Line Plots

sns.relplot(
    data = tips,
    x = "total_bill",
    y = "tip",
    kind = "line",
    hue = "smoker"
)
```
